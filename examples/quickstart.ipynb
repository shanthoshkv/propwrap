```python
from propwrap import Mixture
m = Mixture("RP-1", "LOX")
r = m.evaluate(of=2.56, pc_bar=70, eps=20)
r  # rich table in Jupyter
```

## Find the optimum O/F

```python
sweep = m.scan_of((2.0, 3.2, 0.1), pc_bar=70, eps=20)
sweep.plot(show=True)
sweep.optimum().of_ratio
```

## Size an engine for 1 MN at sea level

```python
r.size(1e6, sea_level=True)
```

## Performance with altitude

```python
from propwrap import altitude_performance
from propwrap.plotting import plot_altitude
rows = altitude_performance(m, 2.56, 70e5, 20, [i * 5000 for i in range(11)])
plot_altitude(rows)
```

## Compare propellants

```python
from propwrap import compare_propellants
compare_propellants(["RP-1/LOX", "CH4/LOX", "LH2/LOX"], pc_bar=70, eps=40)
```
